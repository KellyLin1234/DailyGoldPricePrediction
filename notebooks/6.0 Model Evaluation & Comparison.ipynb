```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = {
    'Model': ['Random Forest', 'Gradient Boosting', 'LSTM', 'Hybrid (RF + LSTM)'],
    'MAE': [775.1643, 922.9305, 2084.1949, 747.2266]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 6))
sorted_mae = df.sort_values(by='MAE', ascending=True)
sns.barplot(x='Model', y='MAE', data=sorted_mae, palette='Blues_r', hue='Model', legend=False)
plt.title('MAE Comparison')
plt.ylabel('MAE Value')

for i, val in enumerate(sorted_mae['MAE']):
    plt.text(i, val, f'{val:.4f}', ha='center', va='bottom')
plt.tight_layout
plt.savefig('mae_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = {
    'Model': ['Random Forest', 'Gradient Boosting', 'LSTM', 'Hybrid (RF + LSTM)'],
    'R2': [0.9974, 0.9968, 0.9788, 0.9975]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 6))
sorted_r2 = df.sort_values(by='R2', ascending=False)
sns.barplot(x='Model', y='R2', data=sorted_r2, palette='Greens', hue='Model', legend=False)
plt.title('R2 Score Comparison')
plt.ylabel('R2 Score')
plt.ylim(0.95, 1.0) 

for i, val in enumerate(sorted_r2['R2']):
    plt.text(i, val, f'{val:.4f}', ha='center', va='bottom')
plt.tight_layout
plt.savefig('R2 Score Comparison_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
